```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files

uploaded = files.upload()


df = pd.read_csv('feature_engineered_cleaned.csv')

print("Shape of original data:", df.shape)
df.head()
```

```python
# Cell 3: OPTIONAL – Sort by time column if needed

# Replace 'date' with your actual datetime column name if you have one.
# If you don't have a datetime column or it's already sorted by time, you can skip this cell.

# Example:
# df['date'] = pd.to_datetime(df['date'])
# df = df.sort_values('date').reset_index(drop=True)

# For now, we'll assume it's already in correct time order
df = df.reset_index(drop=True)

print("Data after optional sorting (showing top 5 rows):")
df.head()
```

```python


n = len(df)
train_end = int(0.7 * n)
val_end = int(0.9 * n)

train_df = df.iloc[:train_end]
val_df   = df.iloc[train_end:val_end]
test_df  = df.iloc[val_end:]

print("Total rows:", n)
print("Train rows:", len(train_df))
print("Validation rows:", len(val_df))
print("Test rows:", len(test_df))
```

```python


train_df = train_df.copy()
val_df = val_df.copy()
test_df = test_df.copy()

train_df['split'] = 'train'
val_df['split']   = 'val'
test_df['split']  = 'test'

dataset_df = pd.concat([train_df, val_df, test_df], axis=0).reset_index(drop=True)

print(dataset_df['split'].value_counts())
dataset_df.head()
```

```python
dataset_df.to_csv('dataset.csv', index=False)
print("dataset.csv saved.")

from google.colab import files
files.download('dataset.csv')
```

```python
train_df.to_csv('train.csv', index=False)
val_df.to_csv('val.csv', index=False)
test_df.to_csv('test.csv', index=False)

print("train.csv, val.csv, and test.csv saved.")

files.download('train.csv')
files.download('val.csv')
files.download('test.csv')
```
